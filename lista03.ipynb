QUESTÃO 01

```python
notas = []

for i in range(5):
  nota = float(input("digite a nota: "))
  notas.append(nota)
soma = 0
for nota in notas:
  somo += nota
media = soma / len(notas)
acima_ou_igual=0
for nota in notas:
   if nota >= media:
    acima_ou_igual += 1
print("notas:" ,notas)
print("media:", media)
print("notas >= media:", acima_ou_igual)
```

QUESTÃO 12

```python
estoque ={
    "arroz": 10,
    "feijao": 8,
    "macarrao": 15,
    "leite": 12
}
produto = input("Digite o produto vendido: ")
quantidade = int(input("Digite a quantidade vendida: "))
if produto in estoque:
    if quantidade <= estoque[produto]:
        estoque[produto] = estoque[produto] - quantidade
        print("Venda realizada!")
    else:
        print("Estoque insuficiente")
else:
    print("Produto não encontrado")
print(estoque)
```

QUESTÃO 15

```python
estoque = []
for i in range(5):
    item = {}
    item["codigo"] = int(input("Código: "))
    item["nome"] = input("Nome: "))
    item["quantidade"] = int(input("Quantidade: "))
    item["preco"] = float(input("Preço: "))
    estoque.append(item)
total = 0
maior = 0
nome_maior = ""
for item in estoque:
    valor = item["quantidade"] * item["preco"]
    total = total + valor
    if valor > maior:
        maior = valor
        nome_maior = item["nome"]
resumo = (len(estoque), total)
print(estoque)
print("Valor total:", total)
print("Maior valor:", nome_maior)
print("Resumo:", resumo)
```
