QUESTÃO 01

```python
notas = []

for i in range(5):
  nota = float(input("digite a nota: "))
  notas.append(nota)
soma = 0
for nota in notas:
  somo += nota
media = soma / len(notas)
acima_ou_igual=0
for nota in notas:
   if nota >= media:
    acima_ou_igual += 1
print("notas:" ,notas)
print("media:", media)
print("notas >= media:", acima_ou_igual)
```
