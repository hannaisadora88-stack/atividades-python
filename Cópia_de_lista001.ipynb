QUESTÃO 05

```python
notas = []
for i in range(2):
    while True:
        try:
            nota = float(input(f"digite a {i+1}ª nota: "))
            if 0 <= nota <= 10:
                notas.append(nota)
                break
            print("a nota deve estar entre 0 e 10")
        except ValueError:
            print("digite uma nota numerica valida")

media = sum(notas) / 2
if media >= 7:
  print("aprovado")
elif media >= 5 :
  print("recuperação")
else:
  print("reprovado")
print("media:", media)

```
