QUESTÃO 15

```python
saldo = 200
entradas = 0
retiradas = 0
recusadas = 0
total_entrada = 0
total_saida = 0

codigo = int(input("Código (1, 2 ou 0): "))

while codigo != 0:

    if codigo == 1:
        valor = float(input("Valor da entrada: "))
        saldo = saldo + valor
        entradas = entradas + 1
        total_entrada = total_entrada + valor

    elif codigo == 2:
        valor = float(input("Valor da retirada: "))

        if valor <= saldo:
            saldo = saldo - valor
            retiradas = retiradas + 1
            total_saida = total_saida + valor
        else:
            print("Saldo insuficiente")
            recusadas = recusadas + 1

    codigo = int(input("Código (1, 2 ou 0): "))

print("Saldo final:", saldo)
print("Entradas:", entradas)
print("Retiradas realizadas:", retiradas)
print("Retiradas recusadas:", recusadas)
print("Total que entrou:", total_entrada)
print("Total que saiu:", total_saida)
```
